```python
import json
import pandas as pd
from pathlib import Path

# Define the path to the summaries file
summaries_file = Path("../data/summaries.jsonl")

# Load the summaries from the JSONL file
summaries = []
with open(summaries_file, "r") as f:
    for line in f:
        if line.strip():
            summaries.append(json.loads(line))

# Convert to DataFrame
df_summaries = pd.DataFrame(summaries)

# Display the first few rows
print(f"Loaded {len(df_summaries)} meeting summaries")
df_summaries.head()
```

```python
import ipywidgets as widgets
from IPython.display import display

# Create widgets for navigation and display
prev_button = widgets.Button(
    description='Previous',
    disabled=False,
    button_style='info',
    tooltip='Go to previous summary',
    icon='arrow-left'
)

next_button = widgets.Button(
    description='Next',
    disabled=False,
    button_style='info',
    tooltip='Go to next summary',
    icon='arrow-right'
)

# Create output widget to display the current summary
output = widgets.Output()

# Create a label to show current position
position_label = widgets.Label(value='Record 1 of {}'.format(len(df_summaries)))

# Create layout for buttons
button_box = widgets.HBox([prev_button, position_label, next_button])

# Variable to track current index
current_idx = 0

# Function to display current summary
def display_summary(idx: int) -> None:
    """
    Display the summary at the given index.
    
    Args:
        idx: Index of the summary to display
    """
    with output:
        output.clear_output()
        if 0 <= idx < len(df_summaries):
            summary = df_summaries.iloc[idx]
            for key, value in summary.items():
                print(f"{key}: {value}")
    
    # Update position label
    position_label.value = f'Record {idx + 1} of {len(df_summaries)}'
    
    # Update button states
    prev_button.disabled = (idx == 0)
    next_button.disabled = (idx == len(df_summaries) - 1)

# Button click handlers
def on_prev_button_clicked(b) -> None:
    """Handle previous button click"""
    global current_idx
    if current_idx > 0:
        current_idx -= 1
        display_summary(current_idx)

def on_next_button_clicked(b) -> None:
    """Handle next button click"""
    global current_idx
    if current_idx < len(df_summaries) - 1:
        current_idx += 1
        display_summary(current_idx)

# Register button click handlers
prev_button.on_click(on_prev_button_clicked)
next_button.on_click(on_next_button_clicked)

# Display the initial summary
display_summary(current_idx)

# Display the widget
display(widgets.VBox([button_box, output]))
```
